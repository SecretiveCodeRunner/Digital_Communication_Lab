## 2. Mathematical Theory & Analytical Derivations

### 2.1 The Detection Problem in Additive White Gaussian Noise (AWGN)
Consider a binary digital transmission system wherein an isolated pulse $s(t)$ of duration $T$ is transmitted across an AWGN channel. The received waveform $r(t)$ is:
$$r(t) = s(t) + n(t), \quad 0 \le t \le T$$
where $n(t)$ is a zero-mean stationary Gaussian white noise process characterized by a two-sided power spectral density (PSD):
$$S_n(f) = \frac{N_0}{2} \quad (\mathrm{W/Hz})$$

The receiver passes $r(t)$ through a causal linear time-invariant (LTI) filter with impulse response $h(t)$ and frequency response $H(f)$. The filter output is:
$$y(t) = r(t) * h(t) = s_o(t) + n_o(t)$$
where $s_o(t) = s(t) * h(t)$ is the deterministic signal component and $n_o(t) = n(t) * h(t)$ is the filtered zero-mean Gaussian noise process.

At the decision instant $t = T$, the signal component is:
$$s_o(T) = \int_{-\infty}^{\infty} H(f) S(f) e^{j 2\pi f T} df$$
The output noise variance $\sigma_o^2 = E[n_o^2(T)]$ is obtained by integrating the output noise PSD over all frequencies:
$$\sigma_o^2 = \int_{-\infty}^{\infty} S_{n_o}(f) df = \int_{-\infty}^{\infty} S_n(f) |H(f)|^2 df = \frac{N_0}{2} \int_{-\infty}^{\infty} |H(f)|^2 df$$

The instantaneous Signal-to-Noise Ratio (SNR) at sampling instant $t = T$ is defined as:
$$\mathrm{SNR}_o = \frac{|s_o(T)|^2}{\sigma_o^2} = \frac{\left| \int_{-\infty}^{\infty} H(f) S(f) e^{j 2\pi f T} df \right|^2}{\frac{N_0}{2} \int_{-\infty}^{\infty} |H(f)|^2 df}$$

### 2.2 Derivation of the Matched Filter via Cauchy-Schwarz Inequality
To find the frequency response $H(f)$ that maximizes $\mathrm{SNR}_o$, we invoke the Schwarz Inequality for complex functions:
$$\left| \int_{-\infty}^{\infty} U(f) V(f) df \right|^2 \le \int_{-\infty}^{\infty} |U(f)|^2 df \int_{-\infty}^{\infty} |V(f)|^2 df$$
with equality holding if and only if $U(f) = k \cdot V^*(f)$ for some arbitrary constant $k$.

Letting $U(f) = H(f)$ and $V(f) = S(f) e^{j 2\pi f T}$, the numerator of $\mathrm{SNR}_o$ satisfies:
$$\left| \int_{-\infty}^{\infty} H(f) S(f) e^{j 2\pi f T} df \right|^2 \le \int_{-\infty}^{\infty} |H(f)|^2 df \int_{-\infty}^{\infty} |S(f)|^2 df$$

Substituting this upper bound into the SNR expression:
$$\mathrm{SNR}_o \le \frac{\int_{-\infty}^{\infty} |H(f)|^2 df \int_{-\infty}^{\infty} |S(f)|^2 df}{\frac{N_0}{2} \int_{-\infty}^{\infty} |H(f)|^2 df} = \frac{\int_{-\infty}^{\infty} |S(f)|^2 df}{\frac{N_0}{2}}$$

By Parseval's theorem, $\int_{-\infty}^{\infty} |S(f)|^2 df = \int_{-\infty}^{\infty} s^2(t) dt = E$, where $E$ is the total signal energy. Thus:
$$\mathrm{SNR}_{o,\max} = \frac{2E}{N_0}$$

The upper bound is achieved with equality when:
$$H_{\mathrm{opt}}(f) = k \cdot S^*(f) e^{-j 2\pi f T}$$
Taking the inverse Fourier transform:
$$h_{\mathrm{opt}}(t) = \mathcal{F}^{-1}\{H_{\mathrm{opt}}(f)\} = k \int_{-\infty}^{\infty} S^*(f) e^{-j 2\pi f T} e^{j 2\pi f t} df = k \int_{-\infty}^{\infty} S^*(f) e^{j 2\pi f (t - T)} df$$
Since for real-valued signals $s(t)$, $\mathcal{F}\{s(t)\} = S(f) \iff \mathcal{F}\{s(-t)\} = S^*(f)$:
$$h_{\mathrm{opt}}(t) = k \cdot s(T - t)$$

**Core Conclusion:** The impulse response of the optimal receiver filter in AWGN is the **time-reversed and delayed version of the transmitted input signal**.

### 2.3 The Energy Invariance Theorem
The maximum attainable output SNR is:
$$\mathrm{SNR}_{\max} = \frac{2E}{N_0}$$
Notice that the expression contains **only the signal energy $E$ and the noise power spectral density $N_0/2$**. The physical pulse waveshape (rectangular, sinusoidal, triangular, or raised-cosine) has **zero impact** on the peak SNR in white Gaussian noise, provided each pulse carries identical total energy $E$.

### 2.4 Matched Filter Output as Deterministic Autocorrelation
The continuous signal component at the output of the matched filter is:
$$s_o(t) = s(t) * h(t) = \int_{-\infty}^{\infty} s(\tau) h(t - \tau) d\tau = \int_{-\infty}^{\infty} s(\tau) s(T - (t - \tau)) d\tau = \int_{-\infty}^{\infty} s(\tau) s(\tau + T - t) d\tau$$
Defining the deterministic autocorrelation function of $s(t)$ as $R_{ss}(	au) = \int_{-\infty}^{\infty} s(t) s(t + \tau) dt$:
$$s_o(t) = R_{ss}(t - T)$$
At the decision instant $t = T$:
$$s_o(T) = R_{ss}(0) = \int_{-\infty}^{\infty} s^2(t) dt = E$$
Because the autocorrelation function of any real signal attains its global maximum at zero lag ($|R_{ss}(\tau)| \le R_{ss}(0)$), the matched filter output reaches its absolute maximum precisely at $t = T$.

### 2.5 Equivalence with the Correlator (Multiplier-Integrator) Receiver
An alternative receiver architecture is the active **correlator receiver**, which multiplies the received signal $r(t)$ by a synchronized local template $s(t)$ and integrates over the symbol interval $[0, T]$:
$$y_{\mathrm{corr}} = \int_0^T r(t) s(t) dt$$

Meanwhile, the matched filter output sampled at $t = T$ is:
$$y_{\mathrm{MF}}(T) = \left[ \int_{-\infty}^{\infty} r(\tau) h(t - \tau) d\tau \right]_{t=T} = \int_0^T r(\tau) s(T - (T - \tau)) d\tau = \int_0^T r(\tau) s(\tau) d\tau$$
Therefore:
$$y_{\mathrm{MF}}(T) \equiv y_{\mathrm{corr}}$$

**Operational Comparison:**
- At sampling instant $t = T$, both receivers produce mathematically and statistically identical decision variables.
- For $t \ne T$, the matched filter continuously computes the sliding correlation over all past inputs, whereas the correlator's integrator ramps up during $[0, T]$ and must be reset (dumped) to zero at each symbol boundary.

### 2.6 Timing Jitter & Sampling Phase Sensitivity
If the receiver sampling clock suffers from a timing error $\Delta t$, the signal is sampled at $t_s = T + \Delta t$. Expanding $s_o(T + \Delta t) = R_{ss}(\Delta t)$ in a Taylor series about $\Delta t = 0$:
$$R_{ss}(\Delta t) \approx R_{ss}(0) + R'_{ss}(0)\Delta t + \frac{1}{2}R''_{ss}(0)\Delta t^2$$
Since $R_{ss}(\tau)$ is an even function, $R'_{ss}(0) = 0$. Thus:
$$s_o(T + \Delta t) \approx E + \frac{1}{2} R''_{ss}(0) \Delta t^2$$
The relative SNR penalty is:
$$\frac{\mathrm{SNR}(T + \Delta t)}{\mathrm{SNR}(T)} \approx \left( 1 - \frac{|R''_{ss}(0)|}{2 E} \Delta t^2 \right)^2 \approx 1 - \frac{|R''_{ss}(0)|}{E} \Delta t^2$$
For rectangular pulses, $R_{ss}(\tau)$ has a discontinuous derivative (sharp triangular peak), producing a severe linear falloff $1 - |\Delta t|/T$. For smooth pulses (half-sine, raised-cosine), $R_{ss}(\tau)$ is rounded with parabolic curvature, providing higher resilience against small timing jitter.

### 2.7 Channel Dispersion & Peak Inter-Symbol Interference (ISI)
When a pulse passes through a band-limited dispersive channel $h_c(t)$, high-frequency spectral components are truncated. The received pulse broadens into $p(t) = s(t) * h_c(t)$, and the matched filter output becomes $g(t) = p(t) * h(t)$.

Sampling a multi-symbol transmission $y(t) = \sum_k a_k g(t - k T)$ at $t = m T$ yields:
$$y(m T) = a_m g(0) + \sum_{k \ne m} a_k g((m - k) T) + v(m T)$$
The Nyquist Peak Distortion is defined as:
$$D_{\mathrm{ISI}} = \frac{1}{|g(0)|} \sum_{k \ne 0} |g(k T)|$$
If $D_{\mathrm{ISI}} < 1$, the eye is open and error-free transmission is possible without noise. If $D_{\mathrm{ISI}} \ge 1$, the eye is completely closed, causing errors even at infinite SNR.

## 5. Viva-Voce Examination Preparation (12 Core Technical Questions & Answers)

1. **What is a Matched Filter, and what optimization criterion does it satisfy?**  
   *Answer:* A matched filter is a linear filter designed to maximize the instantaneous signal-to-noise ratio (SNR) at a predetermined sampling instant $t = T$ in the presence of additive stationary noise.

2. **State the mathematical relationship between the input signal $s(t)$ and the matched filter impulse response $h(t)$.**  
   *Answer:* $h(t) = k \cdot s(T - t)$, where $T$ is the symbol duration and $k$ is an arbitrary gain constant. The impulse response is the time-reversed and delayed version of the transmitted pulse.

3. **State and explain the Energy Invariance Theorem.**  
   *Answer:* In white Gaussian noise, the maximum peak output SNR is $\mathrm{SNR}_{\max} = \frac{2E}{N_0}$. It depends exclusively on total pulse energy $E$ and noise spectral density $N_0/2$, and is completely independent of the signal waveshape.

4. **How is the matched filter output related to the autocorrelation function?**  
   *Answer:* The output signal is given by $s_o(t) = R_{ss}(t - T)$, where $R_{ss}(\tau)$ is the deterministic autocorrelation of $s(t)$. The peak occurs at $t = T$, where $s_o(T) = R_{ss}(0) = E$.

5. **Prove why a Correlator receiver is mathematically equivalent to a Matched Filter at $t = T$.**  
   *Answer:* The correlator calculates $\int_0^T r(t) s(t) dt$. The matched filter convolution sampled at $t = T$ is $[r * h](T) = \int_0^T r(\tau) s(T - (T - \tau)) d\tau = \int_0^T r(\tau) s(\tau) d\tau$. Both evaluate to the exact same integral at $t = T$.

6. **What is the practical difference between a Matched Filter and a Correlator?**  
   *Answer:* The matched filter is a passive LTI filter that operates continuously and requires no external timing except at the sampling strobe. The correlator is a time-variant active system requiring an explicit multiplier and a periodic integrate-and-dump reset circuit synchronized to the symbol clock.

7. **Why do rectangular pulses suffer greater sensitivity to timing jitter than half-sine or raised-cosine pulses?**  
   *Answer:* The autocorrelation of a rectangular pulse has a sharp triangular apex with a discontinuous first derivative, causing a rapid linear drop in amplitude ($|\Delta t|$). Smooth pulses have zero slope at the peak ($R'_{ss}(0) = 0$), yielding a gentle quadratic falloff ($\Delta t^2$).

8. **What causes Inter-Symbol Interference (ISI) in a band-limited channel?**  
   *Answer:* Truncation of high-frequency spectral components causes temporal pulse broadening (dispersion). The extended tails of past and future pulses spill into adjacent symbol slots, interfering with the desired symbol.

9. **What information is obtained from the vertical eye opening in an eye diagram?**  
   *Answer:* The vertical eye height quantifies the **noise margin** of the communication system. A larger opening means the system can tolerate higher additive noise without making decision errors.

10. **What does the horizontal eye opening indicate?**  
    *Answer:* The horizontal eye width indicates the **timing jitter margin**—the permissible range of clock phase offsets over which the signal can be sampled without incurring ISI distortion.

11. **What is Nyquist's Peak Distortion Criterion?**  
    *Answer:* $D_{\mathrm{ISI}} = \frac{1}{|g(0)|} \sum_{k \ne 0} |g(k T)|$. When $D_{\mathrm{ISI}} < 1$, the eye is open. When $D_{\mathrm{ISI}} \ge 1$, the eye is completely closed and bit errors will occur even in the absence of noise.

12. **Why do real-world systems sometimes use sub-optimal RC filters instead of true matched filters?**  
    *Answer:* Sub-optimal RC filters require only a single resistor and capacitor, drastically reducing hardware cost, power consumption, and silicon area in simple low-cost transceivers (e.g., RFID, remote controls), where a $1$–$2\,\mathrm{dB}$ SNR penalty is acceptable.
